```python
!pip install fsspec==2025.3.2 datasets evaluate --quiet
from datasets import load_dataset
```

```python
sharegpt = load_dataset("anon8231489123/ShareGPT_Vicuna_unfiltered", split="train[:20]")

predictions = []
references = []

for example in sharegpt:
    convs = example["conversations"]
    if len(convs) < 2 or convs[0]['from'] != 'human' or convs[1]['from'] != 'gpt':
        continue

    prompt = convs[0]['value']
    expected = convs[1]['value']

    inputs = tokenizer(prompt, return_tensors="pt").to(model_test.device)
    with torch.no_grad():
        output = model_test.generate(**inputs, max_new_tokens=200)

    decoded = tokenizer.decode(output[0], skip_special_tokens=True)
    predictions.append(decoded.strip())
    references.append(expected.strip())
```

```python
def exact_match(p, r):
    return p.strip() == r.strip()

correct = sum(exact_match(p, r) for p, r in zip(predictions, references))
accuracy = correct / len(predictions) if predictions else 0.0

print(f"\nEvaluated {len(predictions)} examples")
print(f"Accuracy (Exact Match): {accuracy * 100:.2f}%")

for i in range(min(3, len(predictions))):
    print(f"\n--- PROMPT ---\n{sharegpt[i]['conversations'][0]['value']}")
    print(f"--- EXPECTED ---\n{references[i]}")
    print(f"--- GENERATED ---\n{predictions[i]}")
```
